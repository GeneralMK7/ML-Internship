# **✅ Task 16: Hyperparameter Tuning using GridSearchCV:**

## **Import Required Modules:**

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

## **Load dataset and apply basic preprocessing:**

```python
data=load_breast_cancer(as_frame=True)
df=data.frame
```

```python
X=df.drop(columns="target")
y=df["target"]
```

```python
numerical_features = X.select_dtypes(include=['float64']).columns
categorical_features = X.select_dtypes(include=['int64']).columns
```

```python
ss = StandardScaler().set_output(transform="pandas")
```

```python
X_scaled=ss.fit_transform(X)
X_scaled
```

## **Split data into training and testing sets:**

```python
X_train,X_test,y_train,y_test=train_test_split(X_scaled,y,test_size=0.2,random_state=42)
```

## **Define parameter grid for tuning and apply GridSearchCV:**

```python
clf = GridSearchCV(SVC(gamma="auto"),{"C":[1,10,20],"kernel":["rbf","linear"]},cv=5,return_train_score=False)
```

```python
clf.fit(X_train,y_train)
```

```python
results = pd.DataFrame(clf.cv_results_)
results
```

```python
results[['params','mean_test_score']]
```

## **Extract best parameters and trained model:**

```python
clf.best_score_
```

```python
clf.best_params_
```

```python
best_model = clf.best_estimator_
y_pred = best_model.predict(X_test)
```

```python
tuned_accuracy = accuracy_score(y_test,y_pred)
print("classification Report of best model" ,classification_report(y_test,y_pred))
```

## **Compare tuned model performance with default model:**

```python
ss = SVC(kernel="linear",C=10)
```

```python
ss.fit(X_train,y_train)
```

```python
y_pred = ss.predict(X_test)
```

```python
default_accuracy = accuracy_score(y_test,y_pred)
print(classification_report(y_test,y_pred))
```

```python
print("accuracy difference is ",(tuned_accuracy-default_accuracy)*100)
```
